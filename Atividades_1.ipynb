4- Diferencie, com um exemplo da tabela, uma variável Quantitativa Discreta de uma Quantitativa Contínua. 

R: Variável quantitativa discreta é aquela que assume apenas valores inteiros. Um exemplo da tabela é a contagem de veículos, pois não existe meio veículo. Os valores possíveis são 10, 11, 12 e assim por diante.
Variável quantitativa contínua é aquela que pode assumir qualquer valor, incluindo casas decimais. Um exemplo da tabela é a velocidade média, que pode ser 60,1 km/h, 60,25 km/h, etc.


##Atividade 3: AED Hands-on - Analisando Salários (60 min)
#R:

```python
salarios = [2000, 2500, 2200, 2800, 2100, 2400, 25000]

# Média
media = sum(salarios) / len(salarios)

# Mediana
salarios_ordenados = sorted(salarios)
mediana = salarios_ordenados[3]

# Resultados
print(f"A media salarial e: R$ {round(media, 2)}")
print(f"A mediana salarial e: R$ {mediana}")
```

1- Qual medida (média ou mediana) representa melhor a realidade da maioria dos funcionários desta empresa? Por quê? 
A mediana representa melhor.
R: Porque a maioria dos salários está entre 2000 e 2800, mas existe um valor muito alto (25000) que puxa a média para cima. A mediana pega o valor do meio e não é afetada por esse extremo, então reflete melhor o que a maioria ganha.

2- O valor de R$ 25.000,00 pode ser considerado um outlier? Como ele afetou o cálculo da média? 
R: Sim, é um outlier. Ele é muito maior que todos os outros salários da lista. Esse valor aumentou bastante a média. Sem ele, a média ficaria próxima de 2333, mas com ele sobe para mais de 5500. Ou seja, distorce a realidade.

3- Se o salário do Diretor (o valor mais alto) subisse para R$ 50.000,00, a mediana sofreria alguma alteração? Justifique. 
R: Não muda. A mediana depende apenas da posição central da lista ordenada. Como o valor mais alto continua sendo o último da lista, o valor do meio (2400) permanece o mesmo.

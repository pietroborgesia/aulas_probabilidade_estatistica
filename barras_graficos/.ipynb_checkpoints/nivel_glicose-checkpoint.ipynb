```python
#nivel de gligose
from collections import Counter
import statistics as st
```

```python
dados = [56, 61, 57, 77, 62, 75, 63, 55, 64, 60,
60, 57, 61, 57, 67, 62, 69, 67, 68, 59,
65, 72, 65, 61, 68, 73, 65, 62, 75, 80, 
66, 61, 69, 76, 72, 57, 75, 68, 83, 64, 
69, 64, 66, 74, 65, 76, 65, 58, 65, 64, 
65, 60, 65, 80, 66, 80, 68, 55, 66, 71]
```

```python
#distribuição de frequência

frequencia = Counter(dados)
sorted(frequencia.items())
```

```python
#medidas de tendência central

#media

media = st.mean(dados)
print(f'A média é {media}')
```

```python
#moda

moda = st.mode(dados)
print(f'A moda é {moda}')
```

```python
moda_s = st.multimode(dados)
print(f'As modas são {moda_s}')
```

```python
#mediana
media = st.median(dados)
print(f'A mediana é {media}')
```

```python
#variância
variancia = st.variance(dados)
print(f'A variância é {variancia}')
```

```python
#desvio padrão
desvio = st.stdev(dados)
print(f'O desvio padrão é {desvio}')
```

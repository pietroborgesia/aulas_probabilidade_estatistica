```python
#importação do pandas para leitura de arquivo .csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# url da base de dados para exemplo
url = "https://raw.githubusercontent.com/Ujeverson/datasets/main/insect.csv"
```

```python
#data frame dos dados .csv
df = pd.read_csv(url)
df
```

```python
df.plot.bar()
```

```python
agrupar = df.groupby(["spray"])["count"].sum()
agrupar
```

```python
#criar um gráfico com 512 x 512
plt.figure(figsize=(5,5))
agrupar.plot.bar(color = "black")
```

```python
agrupar.plot.bar(color = ["blue", "yellow", "red"], title = "Eficácia de inseticidas", xlabel="inseticida", ylabel = "Qtde insetos")
```
